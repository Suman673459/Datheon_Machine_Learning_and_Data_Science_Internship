```python
# Data handling
import pandas as pd
import numpy as np

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# Evaluation
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    roc_curve
)

# Models
from sklearn.ensemble import RandomForestClassifier

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

print("All libraries imported successfully!")
```

```python
import os

# Display available files
print("Files in project folder:")
print(os.listdir())
```

```python
df = pd.read_csv("Telco-Customer-Churn.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Remove customerID because it is not useful for prediction
df.drop("customerID", axis=1, inplace=True)

# Convert TotalCharges into numeric format
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Check missing values after conversion
print("Missing values after conversion:")
print(df.isnull().sum())

# Fill missing TotalCharges with median
df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].median())

print("\nData cleaning completed!")
print("New Dataset Shape:", df.shape)
```

```python
# Count customers in each churn category
print("Customer Churn Distribution:")
print(df["Churn"].value_counts())

# Visualize churn distribution
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Churn")

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

plt.show()
```

```python
churn_percentage = df["Churn"].value_counts(normalize=True) * 100

print("Churn Percentage:")
print(churn_percentage.round(2))
```

```python
# Analyze churn by contract type

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Contract",
    hue="Churn"
)

plt.title("Customer Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="tenure"
)

plt.title("Customer Churn vs Tenure")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")

plt.show()
```

```python
# Separate features and target

X = df.drop("Churn", axis=1)
y = df["Churn"]

# Convert target into numerical values
y = y.map({"No": 0, "Yes": 1})

# Identify categorical and numerical columns

categorical_features = X.select_dtypes(include=["object"]).columns
numerical_features = X.select_dtypes(exclude=["object"]).columns

print("Categorical Features:")
print(list(categorical_features))

print("\nNumerical Features:")
print(list(numerical_features))

print("\nTarget Distribution:")
print(y.value_counts())
```

```python
# Split dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())
```

```python
# Create preprocessing pipeline

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("num", "passthrough", numerical_features)
    ]
)

# Apply encoding
X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)

print("Encoding completed successfully!")

print("Training Data Shape:", X_train_encoded.shape)
print("Testing Data Shape:", X_test_encoded.shape)
```

```python
# Import Random Forest
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42
)

# Train the model
rf_model.fit(X_train_encoded, y_train)

print("Random Forest Model trained successfully!")
```

```python
# Predict customer churn
y_pred_rf = rf_model.predict(X_test_encoded)

# Predict probabilities
y_prob_rf = rf_model.predict_proba(X_test_encoded)[:, 1]

print("Predictions completed successfully!")

print("\nFirst 10 Predictions:")
print(y_pred_rf[:10])
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

# Calculate accuracy
accuracy_rf = accuracy_score(y_test, y_pred_rf)

# Calculate ROC-AUC
roc_auc_rf = roc_auc_score(y_test, y_prob_rf)

print("Random Forest Model Evaluation")
print("--------------------------------")
print("Accuracy:", round(accuracy_rf * 100, 2), "%")
print("ROC-AUC Score:", round(roc_auc_rf, 4))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
```

```python
from xgboost import XGBClassifier

print("XGBoost imported successfully!")
```

```python
# Create XGBoost model

xgb_model = XGBClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=len(y_train[y_train == 0]) / len(y_train[y_train == 1]),
    random_state=42,
    eval_metric="logloss"
)

# Train model
xgb_model.fit(X_train_encoded, y_train)

print("XGBoost Model trained successfully!")
```

```python
# Predict customer churn
y_pred_xgb = xgb_model.predict(X_test_encoded)

# Predict probabilities
y_prob_xgb = xgb_model.predict_proba(X_test_encoded)[:, 1]

print("XGBoost predictions completed!")

print("\nFirst 10 Predictions:")
print(y_pred_xgb[:10])
```

```python
# Calculate metrics
accuracy_xgb = accuracy_score(y_test, y_pred_xgb)
roc_auc_xgb = roc_auc_score(y_test, y_prob_xgb)

print("XGBoost Model Evaluation")
print("------------------------")
print("Accuracy:", round(accuracy_xgb * 100, 2), "%")
print("ROC-AUC Score:", round(roc_auc_xgb, 4))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_xgb))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_xgb))
```

```python
models = ["Random Forest", "XGBoost"]

accuracy_scores = [
    accuracy_rf * 100,
    accuracy_xgb * 100
]

roc_scores = [
    roc_auc_rf,
    roc_auc_xgb
]

plt.figure(figsize=(9, 5))

x = np.arange(len(models))
width = 0.35

plt.bar(x - width/2, accuracy_scores, width, label="Accuracy (%)")
plt.bar(x + width/2, [score * 100 for score in roc_scores],
        width, label="ROC-AUC (%)")

plt.xticks(x, models)
plt.ylabel("Score (%)")
plt.title("Random Forest vs XGBoost Performance")
plt.legend()
plt.show()
```

```python
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)
fpr_xgb, tpr_xgb, _ = roc_curve(y_test, y_prob_xgb)

plt.figure(figsize=(8, 6))

plt.plot(fpr_rf, tpr_rf, label=f"Random Forest (AUC = {roc_auc_rf:.3f})")
plt.plot(fpr_xgb, tpr_xgb, label=f"XGBoost (AUC = {roc_auc_xgb:.3f})")

plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")

plt.legend()
plt.show()
```

```python
# Get feature names after One-Hot Encoding
feature_names = preprocessor.get_feature_names_out()

# Get XGBoost feature importance
importance = xgb_model.feature_importances_

# Create feature importance DataFrame
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
})

# Sort by importance
feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

# Display top 15 features
print("Top 15 Important Features:")
display(feature_importance.head(15))
```

```python
plt.figure(figsize=(10, 7))

sns.barplot(
    data=feature_importance.head(15),
    x="Importance",
    y="Feature"
)

plt.title("Top 15 Features Affecting Customer Churn")
plt.xlabel("Feature Importance")
plt.ylabel("Features")

plt.show()
```
